```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

print("Libraries imported successfully!")
```

Load and clean the dataset

```python
# Load the CSV
df = pd.read_csv(r"D:\ML Projects\Spam Detector\spam.csv", encoding="latin-1")

# Grab only the first two columns (ignores any weird empty columns at the end)
df = df.iloc[:, [0, 1]] 

# Rename them so they are easy to work with
df.columns = ["label", "message"]

# Convert labels to numbers (ham = 0, spam = 1)
df["label"] = df["label"].map({"ham": 0, "spam": 1})

# Display the first 5 rows to confirm it worked!
df.head()
```

```python
# 1. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df["message"], 
    df["label"], 
    test_size=0.2, 
    random_state=42
)

# 2. Initialize the Vectorizer
vectorizer = TfidfVectorizer(stop_words="english", lowercase=True)

# 3. Fit it on the training data and transform the text to numbers
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print(f"Training data shape: {X_train_vec.shape}")
```

Traing the model

```python
# Initialize and train the model
model = LogisticRegression()
model.fit(X_train_vec, y_train)

print("Model training complete!")
```

Evaluate the model

```python
# Make predictions on the test set
predictions = model.predict(X_test_vec)

# Calculate and print the accuracy
accuracy = accuracy_score(y_test, predictions)
print(f"Accuracy: {accuracy * 100:.2f}%\n")
print("Classification Report:")
print(classification_report(y_test, predictions, target_names=["Ham", "Spam"]))
```

```python
def check_message(message):
    # We must vectorize the new message using the exact same vectorizer
    vec_message = vectorizer.transform([message])
    prediction = model.predict(vec_message)
    
    if prediction[0] == 1:
        return "🚨 SPAM Don't Click on any suspicios link"
    else:
        return "✅ NOT SPAM, This is a valid email..!"

# --- Try it out! ---
test_message = "hii team there is a meeting at 11AM today. so be ready for that meeting "
print(f"Message: '{test_message}'")
print(f"Result:  {check_message(test_message)}")
```
